```python
# Qiskit 2.x Example – OpenQASM Demonstration

from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
import qiskit.qasm3 as qasm3  # For OpenQASM 3 export
```

```python
# Create a simple 2-qubit quantum circuit
qc = QuantumCircuit(2)
qc.h(0)
qc.cx(0, 1)
qc.measure_all()
```

```python
print("Quantum Circuit:")
print(qc.draw())
```

```python
# Convert the circuit to OpenQASM 3 format
qasm_code = qasm3.dumps(qc)
print("\nGenerated OpenQASM 3 Code:\n")
print(qasm_code)
```

```python
# Optional: Simulate the circuit
simulator = AerSimulator()
compiled_circuit = transpile(qc, simulator)
result = simulator.run(compiled_circuit).result()
counts = result.get_counts()
print("\nSimulation Result:")
print(counts)
```

```python
#TASK 1
from qiskit import QuantumCircuit

qc = QuantumCircuit(3)

qc.h(0)         # Hadamard on qubit 0
qc.cx(0, 1)     # CNOT from qubit 0 to 1
qc.x(2)         # Pauli-X on qubit 2
qc.barrier()    # Add barrier

qc.draw('text')
```

```python
#TASK 2 & 3

from qiskit.qasm3 import dumps

qasm_code = dumps(qc)
print(qasm_code)
```

```python
#TASK 4
from qiskit import QuantumCircuit
from qiskit.qasm3 import dumps

qc = QuantumCircuit(3)

qc.h(0)
qc.cx(0, 1)
qc.barrier()     # <--- Barrier
qc.x(2)

print(dumps(qc))

```

```python
#TASK 5
from qiskit_aer import Aer

sim = Aer.get_backend('aer_simulator')

# Add measurements to see output
qc.measure_all()

result = sim.run(qc, shots=1024).result()
counts = result.get_counts()

print("Simulation counts:")
print(counts)
```
